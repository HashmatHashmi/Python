### The use of f-string (Using Variables in Strings)
#### Using f-string to compose a message, and then assign the entire message to a variable:

```python
first_name = "Hashmat"
last_name = "Ali"
full_name = f"{first_name} {last_name}" #This strings is called f-strings
print(full_name)
print(f"Hello, {full_name.title()}")
```

### Common String Functions and Methods

**`str.join()`**: Joins elements of a list into a single string using a specified delimiter.


```python
words = ['Hello', 'World']
print(", ".join(words))  # Output: Hello, World
```

**`str.find()`**: Returns the index of the first occurrence of a substring, or -1 if not found.

```python
text = "Hello, World!"
print(text.find("World"))  # Output: 7
print(text.find(","))
```

**`str.replace()`**: Replaces all occurrences of a substring with another substring.

```python
text = "Hello, World!"
print(text.replace("World", "Hashmi"))  # Output: Hello, Python!
```

**`str.count()`**: Returns the number of occurrences of a substring in the string.


```python
text = "Hello, World! Hello!"
print(text.count("Hello"))  # Output: 2
```

**`str.startswith()`**: Checks if the string starts with a specified substring.

```python
text = "Hello, World!"
print(text.startswith("Hello"))  # Output: True
```

**`str.endswith()`**: Checks if the string ends with a specified substring.

```python
text = "Hello, World!"
print(text.endswith("World!"))  # Output: True
```

**`str.isdigit()`**: Returns True if all characters in the string are digits.

```python
text = "12345"
print(text.isdigit())  # Output: True
```

**`str.isalpha()`**: Returns True if all characters in the string are alphabetic.

```python
text = "Hello"
print(text.isalpha())  # Output: True
```

**`str.isalnum()`**: Returns True if all characters in the string are alphanumeric (letters and numbers).

```python
text = "Hello123"
print(text.isalnum())  # Output: True
```

**`str.capitalize()`**: Capitalizes the first character of the string.

```python
text = "hello, world!"
print(text.capitalize())  # Output: Hello, world!
```

**`len()`**: Returns the length of the string.=

```python
text = "Hello, World!"
print(len(text))  # Output: 13
```

**`str.lower()`**: Converts all characters in the string to lowercase.

```python
city = "PESHAWAR"
text = "Hello, World!"
print(text.lower())  # Output: hello, world!.
print(city.lower())
```

**`str.upper()`**: Converts all characters in the string to uppercase.

```python
name = "Hashmi"
text = "Hello, World!"
print(text.upper())  # Output: HELLO, WORLD!
print(name.upper())
```

**`str.title()`**: Converts the first character of each word to uppercase.

```python
text = "hello, world!"
print(text.title())  # Output: Hello, World!
name = "hashmat ali" 
print(name.title())
first_name = "Hashmat"
last_name = "Ali"
full_name = f"{first_name} {last_name}"
message = f"Hello, {full_name.title()}"
print(message)
```

### Stripping WhiteSpaces

The use of strip(), rstrip() and lstrip() methods:

**`str.strip()`**: Removes leading and trailing whitespace. 

```python
text = "   Hello, World!   "
print(text.strip())  # Output: "Hello, World!"
```

**`str.rstrip()`**: Removes trailing whitespace. 

```python
country1 = "Pakistan " # here we give a whiteSpace in the string to the right
country3 = " Dubai " # here we give a whiteSpace in the string to the both the sides
print(country1+country3) # here the white spaces will print as well with all the strings
print(country1.rstrip()) # here we remove the white space to the right of string 
```

**`str.lstrip()`**: Removes leading. 

```python
country2 = " Iran" # here we give a whiteSpace in the string to the left
print(country2.lstrip()) # here we remove the white space to the left of string 
```

### Removing whiteSpaces form right or left or from both sides of the strings are temporirily, so if we want to remove the whitespace from the string permanently, you have to associate the stripped value with the variable name:

```python
favorite_language = 'python ' # length of the string is 7
print(len(favorite_language)) # to see the length of string
favorite_language.rstrip()  # we remove the white space from right side
print(len(favorite_language)) # lenght of the string is still 7
favorite_language = favorite_language.rstrip() # length become 6 permanatly
print(len(favorite_language))
```

### **In the real world, these stripping functions are used most often to clean up user input before it’s stored in a program.**

**`str.split()`**: Splits the string into a list based on a specified delimiter (default is whitespace).

```python
text = "Hello, World!"
print(text.split(","))  # Output: ['Hello', ' World!']
```

### Example of Using Multiple String Functions

```python
text = "   Python is awesome!   "

# Chaining methods
result = text.strip().upper().replace("AWESOME", "GREAT")
print(result)  # Output: PYTHON IS GREAT!
```
